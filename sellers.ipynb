```python
import pandas as pd
```

```python
path = 'data/olist/'
sellers = pd.read_csv(path + "olist_sellers_dataset.csv")
```

```python
sellers
```

```python
sellers.isna().values.any()
```

```python
bins = [0, 30, 60, 100, 200, 300, 2000]
labels = [
    "0–30",
    "31–60",
    "61–100",
    "101–200",
    "201-300",
    "310–2000"
]
```

```python
agrupado = (
    sellers
    .groupby("seller_state")
    .size()
    .reset_index(name="quantidade")
)
agrupado
# agrupado.rename(columns={'quantidade':'quantidade_sellers'}, inplace=True)
# agrupado.cumsum()
# agrupado.cumsum()
# from functools import reduce
# contagem = reduce(lambda x,y: x + y, agrupado.columns)
```

```python
agrupado["faixa"] = pd.cut(
    agrupado["quantidade"],
    bins=bins,
    labels=labels,
    include_lowest=True
)

agrupado
```

```python
from pathlib import Path
import json

arquivo = Path("data/br_states.json")

with arquivo.open("r", encoding="utf-8") as resp:
    br_states = json.load(resp)

import plotly.express as px
fig = px.choropleth_map(agrupado,
                           geojson=br_states,
                           locations='seller_state',
                           featureidkey="id",
                           color="faixa",                         # 🔴 agora é categórico
    color_discrete_sequence=[
        "#eff3ff",  # 0–30
        "#bdd7e7",  # 31–60
        "#6baed6",  # 61–100
        "#3182bd",  # 101–200
        "#08519c"   # 200–2000
    ],
                           map_style="carto-positron",
                            zoom=3,
                            center={"lat": -14.235, "lon": -51.925},
                            opacity=0.7
)

fig.update_layout(margin={"r":0,"t":0,"l":0,"b":0})
fig.show()
```
